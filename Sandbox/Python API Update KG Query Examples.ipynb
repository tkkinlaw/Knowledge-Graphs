### Showing updates to querying at ArcGIS API for Python 2.4.4

```python
from arcgis.gis import GIS
from arcgis.graph import KnowledgeGraph
from pprint import pprint
```

```python
gis = GIS(profile='EntProfile')
gis.users.me.username
```

```python
kgItem = gis.content.get('f709228b2aac4611b14d8cb85ca246e1')
kg = KnowledgeGraph(url=kgItem.url, gis=gis)
q = '''
MATCH (d:Distributor{Material:"Metals"})-[:SellsTo]->(p:ProcessingPlant)
WITH p,COUNT(d) AS distCount
RETURN p, distCount
'''
```

### Original results: Generator
Memory efficient because it only stores one record at a time in memory. The structure is generally a list of dictionaries (depends on what the query returns), which can be difficult to parse through.

```python
res = kg.query_streaming(q)
```

```python
for r in res:
    pprint(r)
```

### New output type: Pandas DataFrame
Spreadsheet-like object. Can be easily exported to a CSV file and used across Python.

```python
newRes = kg.query_streaming(q, as_df=True)
newRes.head(5)
```
